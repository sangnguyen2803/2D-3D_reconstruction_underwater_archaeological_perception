# Sparse 3D confidence and reference trajectory agreement

Read actual downloaded data and completed experiments. No results are fabricated.

```python
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt
PROJECT = Path.cwd()
if PROJECT.name == 'notebooks': PROJECT = PROJECT.parent
```

```python
run = PROJECT/'outputs/visibility_gpu_eval'
cloud = np.load(run/'sparse_confidence.npz')
fig = plt.figure(figsize=(10,7)); ax = fig.add_subplot(projection='3d')
ax.scatter(*cloud['xyz'].T,c=cloud['confidence'],cmap='viridis',s=1,vmin=0,vmax=1); ax.set_title('Heuristic point confidence')
```

```python
metrics = json.loads((run/'metrics.json').read_text()); display(metrics)
plt.hist(cloud['confidence'],bins=40); plt.xlabel('Confidence evidence score'); plt.ylabel('Point count')
```

## Interpretation

Camera poses are published bundle-adjustment references. Dense surface ground truth is unavailable. Assess coverage and reprojection together; confidence is not a calibrated probability.
