# Handcrafted evidence quality under blur

Read actual downloaded data and completed experiments. No results are fabricated.

```python
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt
PROJECT = Path.cwd()
if PROJECT.name == 'notebooks': PROJECT = PROJECT.parent
```

```python
import cv2
from underwater_vision.visibility.deterministic import estimate_visibility
from underwater_vision.utils.io import read_rgb
from underwater_vision.data.dataset_loader import MermaidDataset
from underwater_vision.preprocessing.images import resize_rgb
image, _ = resize_rgb(read_rgb(MermaidDataset(PROJECT/'data/raw/mermaid').frames[0].path))
sharp, cues = estimate_visibility(image)
blurred, _ = estimate_visibility(cv2.GaussianBlur(image, (25,25), 6))
fig, ax = plt.subplots(1,3, figsize=(15,4))
ax[0].imshow(image); ax[1].imshow(sharp,vmin=0,vmax=1); ax[2].imshow(blurred,vmin=0,vmax=1)
print('Mean scores:', sharp.mean(), blurred.mean())
```

```python
plt.hist(sharp.ravel(), bins=50, alpha=.6, label='Original'); plt.hist(blurred.ravel(),bins=50,alpha=.6,label='Blurred'); plt.legend(); plt.title('Reliability sensitivity')
```

## Interpretation

Camera poses are published bundle-adjustment references. Dense surface ground truth is unavailable. Assess coverage and reprojection together; confidence is not a calibrated probability.
