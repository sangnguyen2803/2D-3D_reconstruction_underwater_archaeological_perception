```python
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt
PROJECT = Path.cwd()
if PROJECT.name == 'notebooks': PROJECT = PROJECT.parent
```

```python
from underwater_vision.data.dataset_loader import MermaidDataset
from underwater_vision.utils.io import read_rgb
dataset = MermaidDataset(PROJECT/'data/raw/mermaid')
print('Images:', len(dataset.frames), 'BA reference poses:', len(dataset.poses))
frame = dataset.frames[0]
image = read_rgb(frame.path)
print(frame.name, image.shape)
plt.imshow(image); plt.axis('off')
```

```python
poses = np.array([f.reference_c2w[:3, 3] for f in dataset.frames if f.reference_c2w is not None])
fig = plt.figure(); ax = fig.add_subplot(projection='3d'); ax.plot(*poses.T); ax.set_title('Published BA reference camera centers')
```
